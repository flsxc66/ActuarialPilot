# 03 · 责任准备金评估

演示**将来法**与**Zillmer 法**的责任准备金计算。

```python
from actuarial_pilot.core.life_table import load_default_table
from actuarial_pilot.core.pricing import level_premium
from actuarial_pilot.core.reserve import (
    prospective_reserve,
    retrospective_reserve,
    zillmer_reserve,
)

table = load_default_table("M")
premium = level_premium(100_000, 30, "M", 20, 0.03, table)
print(f"均衡净保费 = ¥{premium:,.2f}")

print("\n各保单年度末的责任准备金：")
print(f"{'年度':<6}{'将来法':>14}{'Zillmer':>14}")
for t in [1, 5, 10, 15, 19]:
    v_pros = prospective_reserve(100_000, 30, "M", 20, 0.03, table, premium, duration=t)
    v_zill = zillmer_reserve(100_000, 30, "M", 20, 0.03, table, premium, 500, duration=t)
    print(f"{t:<6}{v_pros:>14,.2f}{v_zill:>14,.2f}")
```

## 解读

- **将来法**：健康保单的责任准备金随时间单调递增（保费累计 > 赔付与费用）。
- **Zillmer 法**：扣除初始费用后，前几年的准备金更低甚至为负（这是为什么 Zillmer 法对新单销售更友好）。
- 两种方法在末期收敛（剩余期限短，差异消失）。
