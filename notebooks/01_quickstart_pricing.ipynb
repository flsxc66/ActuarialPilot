# 01 · 快速上手：寿险定价

本 Notebook 演示如何用 ActuarialPilot 完成一次**定期寿险定价**。

目标
----
为 **30 岁男性、保额 50 万元、20 年期** 的定期寿险定个价，预定利率 3%。

```python
from actuarial_pilot.core.life_table import load_default_table
from actuarial_pilot.core.pricing import (
    gross_premium,
    level_premium,
    net_single_premium,
)

# 1. 加载生命表
table = load_default_table("M")
print(f"已加载生命表：{table.name}")
print(f"30 岁男性的死亡率 qx = {table.qx(30, 'M'):.6f}")
```

```python
# 2. 产品参数
age = 30
sex = "M"
sum_assured = 500_000   # 50 万元
term = 20                # 20 年期
rate = 0.03              # 预定利率 3%

# 3. 趸交净保费 NSP
nsp = net_single_premium(sum_assured, age, sex, term, rate, table, product="term")
print(f"趸交净保费 NSP = ¥{nsp:,.2f}")

# 4. 均衡年缴净保费
net = level_premium(sum_assured, age, sex, term, rate, table, product="term")
print(f"均衡净保费 = ¥{net:,.2f}")

# 5. 毛保费（含 5% 费用率 + 10% 利润率）
gross = gross_premium(net, expense_load=0.05, profit_margin=0.10)
print(f"毛保费 = ¥{gross:,.2f}")
```

## 解读结果

- **趸交净保费**：客户一次性付清的话，需要 ¥XX 才能覆盖未来 20 年的死亡赔付现值。
- **均衡净保费**：分 20 年缴费、每年 ¥XX。
- **毛保费**：加入公司费用和利润后，客户实际年缴金额。

## 进一步探索

试试改变参数：
- 把 `age` 改成 50 → 保费会显著上升（死亡率上升）
- 把 `rate` 改成 0.04 → 保费会下降（贴现效果）
- 把 `term` 改成 30 → 保费会上升（保障期更长）
